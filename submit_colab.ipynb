```python
import json

with open("predictions.json", encoding="utf-8") as f:
    records = json.load(f)

n = len(records)
print(f"total records: {n}")
for field in ["counterparty", "transaction_method", "processor", "recurring_flag"]:
    present = sum(1 for r in records if r[field].strip())
    print(f"  {field}: present on {present}/{n} ({present/n:.1%})")
```

```python
import requests

SUBMIT_URL = "http://3.108.8.61:8990/submit"

# Fill in your own details before running.
NAME = "YOUR_NAME"
EMAIL = "YOUR_EMAIL"
ROLL_NUMBER = "YOUR_ROLL_NUMBER"
COLLEGE = "YOUR_COLLEGE"


def submit_predictions(predictions_path, name, email, roll_number, college, url=SUBMIT_URL):
    data = {
        "name": name,
        "email": email,
        "roll_number": roll_number,
        "college": college,
    }
    with open(predictions_path, "rb") as f:
        submit_files = {"predictions": ("predictions.json", f, "application/json")}
        response = requests.post(url, data=data, files=submit_files, timeout=120)
    return response


print("About to POST to", SUBMIT_URL)
print("  predictions.json:", n, "records")
print("  name:", NAME, "| email:", EMAIL, "| roll_number:", ROLL_NUMBER, "| college:", COLLEGE)
print("Re-run this cell only after you're sure -- each call spends one of the limited attempts.")
```

```python
# Only run this cell when you're ready to actually submit.
response = submit_predictions("predictions.json", NAME, EMAIL, ROLL_NUMBER, COLLEGE)
print("status:", response.status_code)
try:
    result = response.json()
    print(json.dumps(result, indent=2))
except Exception:
    print(response.text)
```
